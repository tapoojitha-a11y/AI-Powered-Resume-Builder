```python
!pip -q install pymupdf python-docx scikit-learn pandas matplotlib ipywidgets

# ============================================================
# AI-POWERED RESUME–JOB DESCRIPTION MATCHING SYSTEM
# Google Colab - Complete Corrected Version
# ============================================================
#
# FEATURES:
# 1. Resume upload and parsing
# 2. Job description analysis
# 3. AI-style text matching using TF-IDF and cosine similarity
# 4. Match score and skill comparison
# 5. Skill gap analysis
# 6. Resume improvement suggestions
# 7. Dashboard and downloadable reports
#
# NOTE:
# This version does not import transformers, torch, or sympy.
# ============================================================

import re
import io
import warnings
from pathlib import Path

import fitz
import pandas as pd
import matplotlib.pyplot as plt

from docx import Document
from google.colab import files
from IPython.display import display, Markdown

from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity

warnings.filterwarnings("ignore")

print("AI Resume-Job Matching System initialized successfully!")


# ============================================================
# FEATURE 1: RESUME UPLOAD AND PARSING
# ============================================================

display(Markdown("# AI-Powered Resume–Job Description Matching System"))
display(Markdown("## Feature 1: Upload your resume"))

print("Upload a PDF, DOCX, or TXT resume.")

uploaded = files.upload()

if not uploaded:
    raise ValueError("Please upload a resume.")

resume_filename = next(iter(uploaded))
resume_bytes = uploaded[resume_filename]

if len(resume_bytes) > 8 * 1024 * 1024:
    raise ValueError("Please upload a resume smaller than 8 MB.")


def extract_resume_text(filename, content):
    """Extract text from PDF, DOCX, or TXT files."""

    extension = Path(filename).suffix.lower()

    if extension == ".pdf":
        with fitz.open(stream=content, filetype="pdf") as pdf:
            text = "\n".join(page.get_text("text") for page in pdf)
        return text

    elif extension == ".docx":
        document = Document(io.BytesIO(content))

        paragraphs = [
            paragraph.text
            for paragraph in document.paragraphs
        ]

        for table in document.tables:
            for row in table.rows:
                paragraphs.append(
                    " | ".join(cell.text for cell in row.cells)
                )

        return "\n".join(paragraphs)

    elif extension == ".txt":
        return content.decode("utf-8", errors="ignore")

    else:
        raise ValueError(
            "Unsupported format. Upload a PDF, DOCX, or TXT file."
        )


resume_text = extract_resume_text(
    resume_filename,
    resume_bytes
)

if len(resume_text.strip()) < 30:
    raise ValueError(
        "Very little text was extracted from the resume. "
        "If it is a scanned PDF, use OCR or a text-based PDF."
    )

print("\nResume uploaded successfully!")
print("Filename:", resume_filename)
print("Extracted characters:", len(resume_text))


# ============================================================
# FEATURE 2: JOB DESCRIPTION ANALYSIS
# ============================================================

display(Markdown("## Feature 2: Enter the job description"))

print(
    "Paste the complete job description below. "
    "When finished, type END on a separate line."
)

jd_lines = []

while True:
    line = input()

    if line.strip().upper() == "END":
        break

    jd_lines.append(line)

job_description = "\n".join(jd_lines).strip()

if len(job_description) < 20:
    raise ValueError(
        "Please enter a valid job description with sufficient text."
    )

print("\nJob description received successfully!")


# ============================================================
# SKILL DATABASE
# Add more skills here if required.
# ============================================================

SKILL_GROUPS = {
    "Programming Languages": [
        "python", "java", "c++", "c#", "javascript",
        "typescript", "sql", "php", "ruby"
    ],

    "AI and Data Science": [
        "machine learning", "deep learning",
        "artificial intelligence",
        "natural language processing", "nlp",
        "computer vision", "pandas", "numpy",
        "scikit-learn", "tensorflow", "pytorch",
        "data analysis", "data visualization",
        "statistics", "excel", "power bi", "tableau",
        "generative ai", "large language models"
    ],

    "Web Development": [
        "html", "css", "react", "angular", "node.js",
        "flask", "django", "fastapi", "rest api",
        "api development", "web development"
    ],

    "Databases and Cloud": [
        "mysql", "postgresql", "mongodb", "sqlite",
        "oracle", "aws", "azure", "google cloud",
        "cloud computing", "firebase", "redis"
    ],

    "Tools and Software": [
        "git", "github", "docker", "kubernetes",
        "linux", "software development", "software testing",
        "object-oriented programming", "agile", "scrum"
    ],

    "Professional Skills": [
        "communication", "teamwork", "leadership",
        "problem solving", "critical thinking",
        "project management", "time management"
    ]
}

ALL_SKILLS = sorted(
    {skill for group in SKILL_GROUPS.values() for skill in group},
    key=len,
    reverse=True
)


# ============================================================
# SKILL EXTRACTION
# ============================================================

def extract_skills(text):
    """Find known skills in a text document."""

    text = text.lower()
    detected_skills = set()

    for skill in ALL_SKILLS:
        pattern = (
            r"(?<![a-z0-9+#.])"
            + re.escape(skill.lower())
            + r"(?![a-z0-9+#.])"
        )

        if re.search(pattern, text):
            detected_skills.add(skill)

    return detected_skills


resume_skills = extract_skills(resume_text)
job_skills = extract_skills(job_description)


# ============================================================
# FEATURE 3: RESUME MATCHING USING TF-IDF AND COSINE SIMILARITY
# ============================================================

display(Markdown("## Feature 3: AI-based resume matching"))

def calculate_similarity(resume, job_description):
    """
    Compare the resume and job description using TF-IDF
    and cosine similarity. This does not require transformers.
    """

    vectorizer = TfidfVectorizer(
        stop_words="english",
        ngram_range=(1, 2),
        max_features=10000
    )

    try:
        vectors = vectorizer.fit_transform([
            resume,
            job_description
        ])

        score = cosine_similarity(
            vectors[0:1],
            vectors[1:2]
        )[0][0]

        return round(float(score) * 100, 2)

    except ValueError:
        return 0.0


similarity_score = calculate_similarity(
    resume_text,
    job_description
)

print("Text similarity calculated successfully.")
print("TF-IDF similarity score:", similarity_score, "%")


# ============================================================
# FEATURE 4: MATCH SCORE AND SKILL COMPARISON
# ============================================================

display(Markdown("## Feature 4: Match score and skill comparison"))

matched_skills = resume_skills.intersection(job_skills)
missing_skills = job_skills.difference(resume_skills)
additional_skills = resume_skills.difference(job_skills)

if job_skills:
    skill_coverage = (
        len(matched_skills) / len(job_skills)
    ) * 100

    overall_score = (
        0.65 * similarity_score
        + 0.35 * skill_coverage
    )
else:
    skill_coverage = 0.0
    overall_score = similarity_score

overall_score = round(overall_score, 2)
skill_coverage = round(skill_coverage, 2)

print("\n" + "=" * 55)
print("RESUME MATCHING RESULTS")
print("=" * 55)

print(f"Overall Match Score: {overall_score}%")
print(f"Text Similarity Score: {similarity_score}%")
print(f"Required Skill Coverage: {skill_coverage}%")

print("\nMatched Skills:")
print(", ".join(sorted(matched_skills)) or "None detected")

print("\nMissing Skills:")
print(", ".join(sorted(missing_skills)) or "None detected")

print("\nAdditional Skills in Resume:")
print(", ".join(sorted(additional_skills)) or "None detected")


# ============================================================
# FEATURE 5: SKILL GAP ANALYSIS
# ============================================================

display(Markdown("## Feature 5: Skill gap analysis"))

if missing_skills:
    gap_data = []

    for skill in sorted(missing_skills):
        gap_data.append({
            "Missing Skill": skill,
            "Recommended Action": (
                "Review the requirement and develop the skill "
                "if it is relevant to your career goals."
            )
        })

    gap_df = pd.DataFrame(gap_data)
    display(gap_df)

else:
    print(
        "No missing skills from the built-in vocabulary "
        "were detected."
    )

print(
    "\nNote: A missing skill means it was not found in the resume text. "
    "It does not necessarily mean the candidate lacks that skill."
)


# ============================================================
# FEATURE 6: RESUME IMPROVEMENT SUGGESTIONS
# ============================================================

display(Markdown("## Feature 6: Resume improvement suggestions"))

suggestions = []

if missing_skills:
    suggestions.append(
        "Review job-related skills not detected in your resume: "
        + ", ".join(sorted(missing_skills))
        + ". Include them only if you genuinely possess them."
    )

if similarity_score < 50:
    suggestions.append(
        "Tailor your professional summary and work experience "
        "to the target job description."
    )

if not re.search(
    r"\b(project|built|developed|implemented|created|designed)\b",
    resume_text,
    re.IGNORECASE
):
    suggestions.append(
        "Include relevant academic or professional projects "
        "and describe your contribution."
    )

if not re.search(
    r"\b(\d+%|\d+\+?\s?(users|customers|years|projects|records))\b",
    resume_text,
    re.IGNORECASE
):
    suggestions.append(
        "Where accurate, quantify your achievements using "
        "measurable outcomes."
    )

if not re.search(
    r"\b(education|bachelor|degree|university|college|certification)\b",
    resume_text,
    re.IGNORECASE
):
    suggestions.append(
        "Include your education and relevant certifications, "
        "if applicable."
    )

suggestions.append(
    "Use clear headings, consistent formatting, and concise bullet points."
)

suggestions.append(
    "Check spelling and ensure all listed skills accurately reflect "
    "your experience."
)

for number, suggestion in enumerate(suggestions, start=1):
    print(f"{number}. {suggestion}")


# ============================================================
# JOB DESCRIPTION REQUIREMENT EXTRACTION
# ============================================================

display(Markdown("### Extracted job requirements"))

requirement_terms = [
    "require", "qualification", "experience",
    "responsibil", "skill", "proficien",
    "knowledge", "degree", "preferred",
    "must have", "you will"
]

requirement_lines = []

for line in job_description.splitlines():
    cleaned = re.sub(
        r"^[\s•*\-–]+",
        "",
        line
    ).strip()

    if (
        len(cleaned) > 12
        and any(
            term in cleaned.lower()
            for term in requirement_terms
        )
    ):
        requirement_lines.append(cleaned)

if requirement_lines:
    for requirement in requirement_lines[:12]:
        print("-", requirement)
else:
    print(
        "No obvious requirement lines detected. "
        "Review the complete job description."
    )


# ============================================================
# FEATURE 7: DASHBOARD AND VISUALIZATION
# ============================================================

display(Markdown("# Feature 7: Matching dashboard"))

fig, ax = plt.subplots(figsize=(8, 3.5))

labels = [
    "Overall Match",
    "Text Similarity",
    "Skill Coverage"
]

values = [
    overall_score,
    similarity_score,
    skill_coverage
]

bars = ax.barh(labels, values)

ax.set_xlim(0, 100)
ax.set_xlabel("Score (%)")
ax.set_title("Resume–Job Description Matching Dashboard")

for bar, value in zip(bars, values):
    ax.text(
        min(value + 1, 91),
        bar.get_y() + bar.get_height() / 2,
        f"{value:.1f}%",
        va="center"
    )

plt.tight_layout()
plt.show()

summary_df = pd.DataFrame({
    "Metric": [
        "Overall Match Score",
        "Text Similarity",
        "Required Skill Coverage",
        "Matched Skills",
        "Missing Skills",
        "Additional Skills"
    ],

    "Result": [
        f"{overall_score}%",
        f"{similarity_score}%",
        f"{skill_coverage}%",
        len(matched_skills),
        len(missing_skills),
        len(additional_skills)
    ]
})

display(summary_df)


# ============================================================
# GENERATE A DOWNLOADABLE TXT REPORT
# ============================================================

report = f"""
AI-POWERED RESUME–JOB DESCRIPTION MATCHING REPORT
=================================================

Resume File: {resume_filename}

OVERALL MATCH SCORE: {overall_score}%
TEXT SIMILARITY SCORE: {similarity_score}%
REQUIRED SKILL COVERAGE: {skill_coverage}%

MATCHED SKILLS
--------------
{", ".join(sorted(matched_skills)) or "None detected"}

MISSING JOB-RELATED SKILLS
--------------------------
{", ".join(sorted(missing_skills)) or "None detected"}

ADDITIONAL RESUME SKILLS
------------------------
{", ".join(sorted(additional_skills)) or "None detected"}

JOB DESCRIPTION REQUIREMENTS
----------------------------
{chr(10).join("- " + x for x in requirement_lines[:12])}

IMPROVEMENT SUGGESTIONS
-----------------------
{chr(10).join(str(i) + ". " + s for i, s in enumerate(suggestions, 1))}

IMPORTANT NOTE
--------------
This score is an approximate text and skill comparison.
It is not a hiring decision or a guarantee of job suitability.
Skill extraction depends on the built-in skill vocabulary.
"""

report_filename = "resume_job_matching_report.txt"

with open(
    report_filename,
    "w",
    encoding="utf-8"
) as report_file:
    report_file.write(report)

display(Markdown("## Download the matching report"))
files.download(report_filename)


# ============================================================
# DOWNLOAD SKILL COMPARISON AS CSV
# ============================================================

skill_rows = []

for skill in sorted(matched_skills):
    skill_rows.append({
        "Skill": skill,
        "Status": "Matched"
    })

for skill in sorted(missing_skills):
    skill_rows.append({
        "Skill": skill,
        "Status": "Missing from resume text"
    })

for skill in sorted(additional_skills):
    skill_rows.append({
        "Skill": skill,
        "Status": "Additional resume skill"
    })

skills_df = pd.DataFrame(
    skill_rows,
    columns=["Skill", "Status"]
)

csv_filename = "resume_skill_comparison.csv"

skills_df.to_csv(
    csv_filename,
    index=False
)

files.download(csv_filename)

print("\n" + "=" * 55)
print("PROJECT COMPLETED SUCCESSFULLY!")
print("=" * 55)

print("Features implemented: 7/7")
print("Report generated:", report_filename)
print("Skill comparison generated:", csv_filename)
```
